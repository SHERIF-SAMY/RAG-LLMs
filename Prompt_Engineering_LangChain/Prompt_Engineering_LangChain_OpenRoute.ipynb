```python
from transformers import LlamaTokenizer, LlamaForCausalLM, GenerationConfig, pipeline
from langchain.llms import HuggingFacePipeline
import torch
```

```python
# model_id = "meta-llama/Llama-2-7b-chat-hf"
# model_id = "chavinlo/alpaca-native"

# Tokenizer = LlamaTokenizer.from_pretrained(model_id)

# base_model = LlamaForCausalLM.from_pretrained(
#                                               model_id,
#                                               load_in_4bit=True,
#                                               device_map = 'auto'
#                                               )

from transformers import LlamaTokenizer, AutoModelForCausalLM
from transformers import BitsAndBytesConfig

model_id = "chavinlo/alpaca-native"

# Load tokenizer
tokenizer = LlamaTokenizer.from_pretrained(model_id)

# Define quantization config
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",  # or "fp4"
    bnb_4bit_compute_dtype=torch.float16  # or bfloat16
)

# Load model with 4-bit quantization
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto"
)
```

```python
hf_pipeline = pipeline(
    task="text-generation",
    model=model,
    tokenizer=tokenizer,
    max_length=256,
    temperature=0.7,
    top_p=0.95,
)
```

```python
llm = HuggingFacePipeline(pipeline=hf_pipeline)
```

```python
prompt = """
what is computer vision.
""".strip()
```

```python
print(llm(prompt))
```

```python
!pip install -q cohere

from langchain.text_splitter import NLTKTextSplitter
# import nltk
# nltk.download('punkt_tab')

from langchain.document_loaders import WikipediaLoader
# from langchain.text_splitter import NLTKTextSplitter
from langchain.embeddings import CohereEmbeddings
from langchain.vectorstores import Chroma
from transformers import AutoTokenizer
from langchain.text_splitter import CharacterTextSplitter

cohere_api_key = '4IEJ8wtAYOLLuttaFst5G7Ipyx8G4W87kgcgGuLQ'
query = "Nikola Tesla"
model_id = 'stabilityai/stablelm-tuned-alpha-3b'

# Load document
docs = WikipediaLoader(query=query, load_max_docs=1, doc_content_chars_max=20000).load()
document = [docs[0].page_content]
metadatas = [{"document": query}]

# Split document
# text_splitter = NLTKTextSplitter(chunk_size=300, chunk_overlap=50)
# tokens_chunks = text_splitter.create_documents(document, metadatas=metadatas)
#-------

tokenizer = AutoTokenizer.from_pretrained(model_id)

text_splitter = CharacterTextSplitter.from_huggingface_tokenizer(
    tokenizer,
    chunk_size=300,
    chunk_overlap=50,
)

tokens_chunks = text_splitter.create_documents(document, metadatas=metadatas)

# Embedding
embedding_llm = CohereEmbeddings(cohere_api_key=cohere_api_key, user_agent="langchain")

# Vector DB
save_to_dir = "/content/wiki_chroma_db"
vector_db = Chroma.from_documents(
    documents=tokens_chunks,
    embedding=embedding_llm,
    persist_directory=save_to_dir,
)
```

```python
qna_template = "\n".join([
    "Answer the next questionu using the provided context.",
    "If the answer is not contained in the context, say 'NO ANSWER IS AVAILABLE'",
    "### Context:",
    "{context}",
    "",
    "### Question:",
    "{question}",
    "",
    "### Answer:",
])

qna_prmpt = PromptTemplate(
    template=qna_template,
    input_variables=["context", "question"],
    verbose=True
)
```
